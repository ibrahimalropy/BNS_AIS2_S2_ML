```python
email="###!!@mocleW EPGTQ!!!6789"
clean_email=email.strip("#!@123456789")
clean_email
```

```python
Reversethefirstword=clean_email.split()[0]
correct_word=Reversethefirstword[::-1]
correct_word
```

```python
vowels=['a','e','i','o','u']
second_word=clean_email.split()[1]
new_word=""
for i in second_word.lower():
    if i in vowels:
        i.replace(i," ")
    else:
        new_word+=i
final_result=new_word.upper()
```

```python
print(f"Final decoded message:\"{correct_word} {final_result}\"")
```
