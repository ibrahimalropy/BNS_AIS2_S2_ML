```python
email="&&&**$gnirtS PLIO!!@1234"
clean_email=email.strip("@!#$%^&*1234567890")
clean_email
```

```python
first_word=clean_email.split()[0]
correct_word=first_word[::-1]
correct_word
```

```python
second_word=clean_email.split()[1]
replca_char=second_word.replace("I","E").replace("O","U")
replca_char
```

```python
print(f"Final decoded message: {correct_word} {replca_char}")
```
