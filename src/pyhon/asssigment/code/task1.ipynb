```python
email="Amit_ml@gmail.edu"

for i in email.split():
    if i.count("@")>1:
        print("invalid email")
    elif i.count(".")>1:
        print("invalid email")
    elif i.index("@")>i.index("."):
        print("invalid email")
    else:
        print("valid email")            
```

```python
username=email[:email.index("@")]
print(f"Username: {username}")
```

```python
Domain=email[email.index("@")+1:email.index(".")]
print(f"Domain: {Domain}")
```

```python
DomainEnding=email[email.index(".")+1:]
if DomainEnding=="com":
    print("Commercial Domain")
elif DomainEnding=="edu":
    print("Educational Domain")
else:
    print("Other Domain")        
```
