```python
class number:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2


class calculator(number):

    def add(self):
        print("the process add")
        return self.num1 + self.num2

    def subtract(self):
        print("the process subtract")
        return self.num1 - self.num2

    def multiply(self):
        print("the process multiply")
        return self.num1 * self.num2

    def divide(self):
        print("the process divide")
        return self.num1 // self.num2



c1 = calculator(20, 5)


print("Welcome to the calculator!")
print("Choose an operation:")
print("1: Add")
print("2: Subtract")
print("3: Multiply")
print("4: Divide")

choice = input("Enter your choice: ")

if choice == "1":
    print(c1.add())

elif choice == "2":
    print(c1.subtract())

elif choice == "3":
    print(c1.multiply())

elif choice == "4":
    print(c1.divide())

else:
    print("Invalid choice!")
```
