```python
class Task4:
    def __init__(self, number):

        self.number = number
        
    def check_even_odd(self):
        if self.number % 2 == 0:
            return 'Even'
        else:
            return 'Odd'

my_number_task = Task4(14)

result = my_number_task.check_even_odd()

print(f"The number {my_number_task.number} is: {result}")
```
