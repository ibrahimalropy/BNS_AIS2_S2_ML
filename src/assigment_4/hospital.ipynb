```python
class patient :
   def __init__(self ,patientid,name,age,disease):
      self.patientid=patientid
      self.name=name
      self.age=age
      self.disease=disease
class hospital:
   def __init__(self):
    self.patients=[]
   def addpatient (self):
      patientid=int(input("enter id of patient : "))
      name=input("enter name of patient : ")
      age=int(input("enter age of patient : "))
      disease=input("enter disease of patient: ")
      newpa=patient(patientid,name,age,disease)
      self.patients.append(newpa) 
      print("patient is added !")   
   def viewpatient(self):
      if not self.patients:
         print("no patient avaliable ") 
      else:
         for p in self.patients:
             print(f"patient id= {p.patientid}")
             print(f"name of patient= {p.name}")   
             print(f"age of patient= {p.age}")
             print(f"disease= {p.disease}")
             print("-"*30)
   def search(self):
      patientid=int(input("enter id of patient to searche : "))                  
      for p in self.patients:
         if p.patientid==patientid:
             print("-"*30)
             print(f"patient id= {p.patientid}")
             print(f"name of patient= {p.name}")   
             print(f"age of patient= {p.age}")
             print(f"disease= {p.disease}")
             print("-"*30)
         else:
            print ("is not found ") 
   def daletpatient(self):
      patientid=int(input("enter id of patient to delet : "))
      for p in self.patients:
         if p.patientid==patientid:
            self.patients.remove(p)
            print("the delet patient is succcessufully")
            break
         else:
            print("the id is not found to delet ")
h=hospital()
while True:
   print("1- add patient ")                        
   print("2- view patient ")
   print("3- search patient")
   print("4- delet patient " )
   print("5- quit ")
   choose =int(input("enter your choise: "))
   if choose==1:
      h.addpatient()
   elif choose==2:
      h.viewpatient()
   elif choose==3:
      h.search()
   elif choose==4:
      h.daletpatient()
   elif choose==5:
        print(" exiting the app")
        break
   else:
      print("out of range ")                               
```
