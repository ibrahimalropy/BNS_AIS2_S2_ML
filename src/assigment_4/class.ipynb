```python
class Shape:
    def calculate_area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * (self.radius) ** 2


class square(Shape):
    def __init__(self, side):
        self.side = side

    def calculate_area(self):
        return self.side**2


x = Circle(7)
y = square(10)

print("Area: ", x.calculate_area())
print("Area: ", y.calculate_area())
```
