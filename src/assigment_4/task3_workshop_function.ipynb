```python
def add(num1, num2):
    print("the process add")
    return num1 + num2

def subtract(num1, num2):
    print("the process subtract")
    return num1 - num2

def multiply(num1, num2):
    print("the process multiply")
    return num1 * num2

def divide(num1, num2):
    print("the process divide")
    return num1 // num2

n1 = 20
n2 = 5

print("Welcome to the calculator!")
print("Choose an operation:")
print("1: Add")
print("2: Subtract")
print("3: Multiply")
print("4: Divide")

choice = input("Enter your choice: ")

if choice == "1":
    print(add(n1, n2))

elif choice == "2":
    print(subtract(n1, n2))

elif choice == "3":
    print(multiply(n1, n2))

elif choice == "4":
    print(divide(n1, n2))

else:
    print("Invalid choice!")
```
