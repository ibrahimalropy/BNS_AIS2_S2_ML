```python
import pandas as pd
import numpy as np

# Create a synthetic Company Sales Dataset
data = {
    'Transaction_ID': range(1, 11),
    'Product_Category': ['Electronics', 'Home', 'Electronics', 'Sports', 'Home', 
                         'Electronics', 'Home', 'Sports', 'Electronics', 'Electronics'],
    'Sales_Amount': [150, 200, 155, 300, 210, 180, 205, 1000, 190, 160], # 1000 is an Outlier
    'Customer_Age': [25, 34, np.nan, 45, 23, 31, 29, np.nan, 38, 40],    # Contains Nulls (NaN)
    'Rating': [5, 4, 3, 5, 2, 4, 5, 2, 4, 3]
}

df_test = pd.DataFrame(data)

# Save to CSV for students to practice loading files [cite: 74]
df_test.to_csv('company_sales_test.csv', index=False)
print("Test dataset created successfully!")
```

```python
def automated_stat_analyzer(df, column_name):
    if df[column_name].dtype == 'object':
        return {
            'Column': column_name,
            'Mode': df[column_name].mode()[0]
        }
    else:
        return {
            'Column': column_name,
            'Mean': df[column_name].mean(),
            'Median': df[column_name].median(),
            'Std': df[column_name].std()
        }


print("1. Analysis for Sales_Amount:")
sales_stats = automated_stat_analyzer(df_test, 'Sales_Amount')
for key, value in sales_stats.items():
    print(f"{key}: {value}")

print("\n----------------------------------\n")

print("2. Analysis for Product_Category :")
category_stats = automated_stat_analyzer(df_test, 'Product_Category')
for key, value in category_stats.items():
    print(f"{key}: {value}")
```

```python
import pandas as pd

def null_handling_strategy(df, strategy="fill_median"):
    if strategy == "fill_median":
        df['Customer_Age'] = df['Customer_Age'].fillna(df['Customer_Age'].median())
        
    return df

print("Before cleaning:")
print(df_test['Customer_Age'])

df_cleaned = null_handling_strategy(df_test)

print("\nAfter cleaning:")
print(df_cleaned['Customer_Age'])
```
